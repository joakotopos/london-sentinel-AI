Leyendo los datos

```python
# Celda 1: Importación de librerías
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Celda 2: Configuración de rutas (Siguiendo la modularidad del proyecto)
BASE_DIR = os.path.dirname(os.getcwd()) # Sube un nivel desde la carpeta 'notebooks'
PATH_PROCESSED = os.path.join(BASE_DIR, 'data', 'processed', 'london_crime_cleaned.parquet')

# Celda 3: Carga de datos procesados
# Aquí leemos el resultado de la Fase 2: Limpieza y Transformación
df = pd.read_parquet(PATH_PROCESSED)
print(f"Dataset cargado con {len(df)} registros para análisis.")
display(df.head())

# Celda 4: Gráfico de Crímenes por Año (Tendencia Temporal)
# Esto ayuda a validar semánticamente si los datos tienen sentido lógico
plt.figure(figsize=(12, 6))
df_yearly = df.groupby('year')['value'].sum().reset_index()
sns.lineplot(data=df_yearly, x='year', y='value', marker='o', color='teal')
plt.title('Evolución Total de Crímenes en Londres (Dataset Procesado)')
plt.xlabel('Año')
plt.ylabel('Cantidad de Incidentes')
plt.grid(True)
plt.show()

# Celda 5: Top 10 de Categorías Principales (Major Categories)
# Útil para entender la distribución semántica de los datos
plt.figure(figsize=(10, 6))
df_category = df.groupby('major_category')['value'].sum().sort_values(ascending=False).head(10)
df_category.plot(kind='barh', color='skyblue')
plt.title('Top 10 Categorías de Crimen más Frecuentes')
plt.xlabel('Número de Reportes')
plt.ylabel('Categoría')
plt.gca().invert_yaxis()
plt.show()
```
